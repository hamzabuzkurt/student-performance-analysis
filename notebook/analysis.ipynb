## Load the student mathematics dataset

```python
# Load the student mathematics dataset

import pandas as pd

students = pd.read_csv(
    "../data/student-mat.csv",
    sep=";"
)

# Display first 5 records
students.head()
```

## Import libraries first
## Create figure
## Create histogram
## Add title and labels
## Display graph

```python

# Import libraries first
import matplotlib.pyplot as plt
import seaborn as sns

# Create figure
plt.figure(figsize=(10, 5))

# Create histogram
sns.histplot(
    data=students,
    x="G3",
    bins=20,
    kde=True
)

# Add title and labels
plt.title("Distribution of Student Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

# Save graph
plt.savefig(
    "../charts/grade_distribution.png",
    dpi=300,
    bbox_inches="tight"

)

# Display graph
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    data=students,
    x="sex",
    y="G3",
    hue="sex",
    legend=False
)

plt.title("Average Mathematics Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

# Save graph
plt.savefig("../charts/gender_performance.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=students,
    x="studytime",
    y="G3"
)

plt.title("Study Time vs Student Performance")
plt.xlabel("Weekly Study Time Category")
plt.ylabel("Average Final Grade")

plt.savefig("../charts/studytime_vs_grades.png", dpi=300, bbox_inches="tight")

plt.show()
```

## Select important numerical columns
## Calculate correlation

```python
# Select important numerical columns
selected_columns = [
    "age", "studytime", "failures",
    "absences", "G1", "G2", "G3"
]

# Calculate correlation
correlation = students[selected_columns].corr()

# Calculate correlation
correlation
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Student Performance Factors")

plt.savefig("../charts/correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
print("Average First Period Grade (G1):", students["G1"].mean())
print("Average Second Period Grade (G2):", students["G2"].mean())
print("Average Final Grade (G3):", students["G3"].mean())

print("\nCorrelation with Final Grade (G3):")
print(correlation["G3"].sort_values(ascending=False))
```

## Average final grade by number of past failures
## Average final grade by number of absences

```python
# Average final grade by number of past failures
print("Average Final Grade by Past Failures:")
print(students.groupby("failures")["G3"].mean().round(2))

# Average final grade by number of absences
print("\nAverage Final Grade by Absences:")
print(students.groupby("absences")["G3"].mean().round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=students,
    x="failures",
    y="G3"
)

plt.title("Past Failures vs Final Grades")
plt.xlabel("Number of Past Class Failures")
plt.ylabel("Average Final Grade")
plt.savefig("../charts/past_failures.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Analysis of Student Absences and Final Grades

```python
# Average final grade based on student absences

print("Average Final Grade by Absences:")

print(
    students.groupby("absences")["G3"]
    .mean()
    .round(2)
)
```

### Visualization: Absences vs Final Grades

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=students,
    x="absences",
    y="G3"
)

plt.title("Student Absences vs Final Grades")
plt.xlabel("Number of Absences")
plt.ylabel("Average Final Grade")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../charts/absences_vs_grades.png", dpi=300, bbox_inches="tight")
plt.show()
```
